Q2 在 Python 中，用尽量多的不同方式实现求 x^n，其中，n 为正整数，并写出代码。


```python

# 用运算符**
def pow_op(x, n):
    return x ** n

x, n = 2, 9
print("pow_op       :", pow_op(x, n))

# for 循环
def pow_for(x, n):
    result = 1
    for _ in range(n):
        result *= x
    return result

x, n = 2, 9
print("pow_for       :", pow_for(x, n))

# while 循环
def pow_while(x, n):
    result = 1
    k = n
    while k > 0:
        result *= x
        k -= 1
    return result

x, n = 2, 9
print("pow_while       :", pow_while(x, n))
```

Q3  用尽量简短的 Python 代码合并某两个无重复元素列表 A、B 中的元素，并删除合并后所有重复出现的元素。

```python
def merge_unique(A, B):
    result = []
    for x in A + B:
        if x not in result:
            result.append(x)
    return result

A = [2, 3, 5, 4, 1]
B = [5, 7, 8, 3, 2]

print(merge_unique(A, B)) #[2, 3, 5, 4, 1, 7, 8]
```

Q4 某企业销售部门有多名职工，各职工分别存有如下信息： 
职工号、姓名、性别、年龄、职级、工资、工龄、销售额 
使用二重列表clerks 存放，每个元素一名职工，示例数据如下： 
clerks = [ 
['033164', '小明', '男', 30, 'P12', 14239, 3, 52],  
['021796', '小红', '女', 28, 'P09', 36807, 5, 95],  
#...... 其他职工的信息，略 
] 
(1) 完成两个参数
．．．．的比较函数 compare，用于排序时通过模块 functools 
中 cmp_to_key 函数接受，排序准则为按销售额降序、销售额相等则按工龄
递增，该函数在排序时使用格式如下： 
sorted(clerks, key = cmp_to_key(compare)) 
(2) 完成排序时用于比较的单参数
．．． lambda 表达式，排序准则为按工资降
序、工资相等时按工龄递增，该 lambda 表达式在排序时使用格式如下，其
中的“****” 为待完成的lambda 表达式： 
sorted(clerks, key = lambda ****) 

```python
from functools import cmp_to_key

clerks = [
    ['033164', '小明', '男', 30, 'P12', 14239, 3, 52],
    ['021796', '小红', '女', 28, 'P09', 36807, 5, 95],
    ['067532', '小李', '男', 35, 'P11', 36807, 2, 95],
    ['098542', '小王', '女', 26, 'P08', 14239, 1, 52]
]

#(1)
def compare(a, b):
    if a[7] != b[7]:    
        return b[7] - a[7]
    return a[6] - b[6]   
result1 = sorted(clerks, key=cmp_to_key(compare))
print("排序结果 1（销售额降序，工龄升序）:")
for r in result1:
    print(r)

#(2)
result2 = sorted(clerks, key=lambda x: (-x[5], x[6]))
print("\n排序结果 2（工资降序，工龄升序）:")
for r in result2:
    print(r)
```

Q5编写函数check(username, password, cypher) 用于检测用户输入的登录信
息，如果用户名和密码都正确则返回True，否则返回False： 
(1) username 为用户名的字符串，长度4 ~ 12 个字符，只能由英文字母、
数字和下划线组成，第一个字符只能是英文字母； 
(2) password 为密码的字符串，长度6 ~ 16 位字符，只能由英文字母、
数字和特殊字符“+-*/@#$%”组成； 
(3) cypher 为用户名做关键字的字典，对应的密码为键值； 
(4) 用户名和密码均区分大小写。 
在完成函数的基础上实现程序：输入密码文件的“全路径+文件名”， 读 入文
件后建立字典cypher，再调用该函数，检验输入的用户名和密码是否正确，当格
式与用户名和密码完全匹配正确，显示“登录成功，” 否则显示“登录失败”， 失
败 时 可以累计尝试输入三次。 
密码文件为已预存的密码相关信息，每行存放一个用户名与密码，中间
用空格分开。

```python
import re

def check(username, password, cypher):
    """检查用户名密码是否匹配，返回 OK / USER_ERR / PWD_ERR"""

    if username not in cypher:
        return "USER_ERR"

    if not re.fullmatch(r"[A-Za-z0-9+\-*/@#$%]{6,16}", password):
        return "PWD_ERR"

    if cypher[username] != password:
        return "PWD_ERR"

    return "OK"



filepath = input("请输入密码文件路径：").strip().replace("\\", "/")

cypher = {}

try:
    with open(filepath, "r", encoding="utf-8") as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            user, pwd = line.split()
            cypher[user] = pwd

except FileNotFoundError:
    print(" 文件不存在，请检查路径！")
    raise SystemExit


print("\n系统读取到的用户：", list(cypher.keys()), "\n")



for attempt in range(1, 4):
    print(f"第 {attempt} 次尝试登录：")

    username = input("用户名：").strip()

 
    if username not in cypher:
        print(" 用户名不存在！")
        continue

    password = input("密码：").strip()

    result = check(username, password, cypher)

    if result == "OK":
        print(" 登录成功！")
        break
    else:
        print(" 密码错误！")

else:
    print("\n 密码连续错误三次，系统已锁定。")
```

Q6完成一个敏感词过滤函数Filter(origin, result)，分别使用 origin 和 result 
传递进待过滤的“全路径+文件名”与保存过滤结果的“全路径+文件名”，
将 待过滤文件中单独的敏感词
．．．．．．转换为等长度的“*”字符。 
设待过滤文件和敏感词均为纯英文，需要过滤的敏感词存放在默认路径
下的文件“sensitive.txt”中，每行一个敏感词。 
所谓单独的单词指的是前后可以有标点符号、数字和空格，但是不能有
字母，另外，待过滤的敏感词并不区分大小写。

```python
import re
import os

def Filter(origin):
  


    sensitive_file = "sensitive.txt"
    if not os.path.exists(sensitive_file):
        print(" 未找到 sensitive.txt，请确保它与程序在同一目录！")
        return

   
    with open(sensitive_file, "r", encoding="utf-8") as f:
        sensitive_words = [line.strip() for line in f if line.strip()]

   
    try:
        with open(origin, "r", encoding="utf-8") as f:
            text = f.read()
    except FileNotFoundError:
        print(" 找不到 origin 文件，请检查路径！")
        return

    for word in sensitive_words:
        pattern = rf"(?i)(?<![A-Za-z]){re.escape(word)}(?![A-Za-z])"
        replacement = "*" * len(word)
        text = re.sub(pattern, replacement, text)

  
    base, ext = os.path.splitext(origin)
    result = f"{base}_filtered{ext}"


    with open(result, "w", encoding="utf-8") as f:
        f.write(text)

    print(f" 过滤完成！输出文件为：{result}")
```

Q7 某校学生参加某项全国竞赛后，获奖证书丰厚。竞赛主办方为简化流程，
发放的全部是jpg格式的以证书编号为名字的电子证书，且相关的指导教
师与被指导学生共享同一份证书，为方便归档管理，需要将原始证书分别
按参赛学生名与指导教师名复制出来并更名。 
全校学生全国赛获奖证书的相关信息现已存放在名为“C:\竞赛证书\全国
赛证书编号.xls”的 Excel 工作簿文件的“Sheet1”工作表中，该工作表中包
括了证书编号、学生姓名、学生学号、获奖级别、指导教师等信息，示例数
据如下图所示： 
待处理证书均存放在该文件夹下的“原始证书”子文件夹中，为方便证书管
理，读入此 Excel 文件，不改动原始证书时，在文件夹“C:\竞赛证书\”下，另
新建2个文件夹“学生证书”和“教师证书”，各自存放的文档名格式如下： 
(1) 学生证书文件夹：学生姓名-学号-奖励项目.jpg 
(2) 教师证书文件夹：教师姓名-学生名-奖励级别.jpg 
补充说明： 
设所有获奖的学生无重名，每人只有一份证书，所有获奖的指导教师也
无重名。 
因教师可能同时指导多名学生，超过1份证书时需要在教师名后附上01、
02 等数字区分，序号按原始xls文件中记录出现的次序连续递增编号。例如
按照图例中的数据就应该生成“姜维 01-林冲-一等奖.jpg、姜维 02-鲁智深
二等奖.jpg”等证书，但如果指导教师总共只有1份证书则不必加上数字； 
学生证书和教师证书均为原始证书的副本。

```python
import os
import shutil
import xlrd   


BASE_DIR = r"E:\竞赛证书"                   
EXCEL_PATH = os.path.join(BASE_DIR, "全国赛证书编号.xls")
SRC_DIR = os.path.join(BASE_DIR, "原始证书")   

STU_DIR = os.path.join(BASE_DIR, "学生证书")
TEA_DIR = os.path.join(BASE_DIR, "教师证书")


os.makedirs(STU_DIR, exist_ok=True)
os.makedirs(TEA_DIR, exist_ok=True)


def cell_to_str(cell):
    """把单元格内容转成字符串（处理数字变成 float 的情况）"""
    if isinstance(cell, float):
 
        return str(int(cell))
    return str(cell).strip()


def main():

    wb = xlrd.open_workbook(EXCEL_PATH)
    sheet = wb.sheet_by_name("Sheet1")

    rows = []          
    teacher_counts = {}  

  
    for i in range(1, sheet.nrows):
        cert_no = cell_to_str(sheet.cell_value(i, 1))  
        stu_name = cell_to_str(sheet.cell_value(i, 2)) 
        stu_id = cell_to_str(sheet.cell_value(i, 3))   
        award = cell_to_str(sheet.cell_value(i, 4))    
        teacher = cell_to_str(sheet.cell_value(i, 5))  

        rows.append({
            "cert_no": cert_no,
            "stu_name": stu_name,
            "stu_id": stu_id,
            "award": award,
            "teacher": teacher
        })

        teacher_counts[teacher] = teacher_counts.get(teacher, 0) + 1

   
    teacher_index = {t: 0 for t in teacher_counts}

    for row in rows:
        cert_no = row["cert_no"]
        stu_name = row["stu_name"]
        stu_id = row["stu_id"]
        award = row["award"]
        teacher = row["teacher"]

        src_file = os.path.join(SRC_DIR, cert_no + ".jpg")

        if not os.path.exists(src_file):
            print(f"⚠ 找不到原始证书文件：{src_file}")
            continue


        stu_filename = f"{stu_name}-{stu_id}-{award}.jpg"
        stu_path = os.path.join(STU_DIR, stu_filename)
        shutil.copy2(src_file, stu_path)

      
        total = teacher_counts[teacher]
        if total == 1:
           
            teacher_name_part = teacher
        else:
           
            teacher_index[teacher] += 1
            idx = teacher_index[teacher]
            teacher_name_part = f"{teacher} {idx:02d}"

        tea_filename = f"{teacher_name_part}-{stu_name}-{award}.jpg"
        tea_path = os.path.join(TEA_DIR, tea_filename)
        shutil.copy2(src_file, tea_path)

        print(f"已处理：{cert_no} -> {stu_filename} / {tea_filename}")

    print("全部证书处理完成！")


if __name__ == "__main__":
    main()
```
